```python
from typing import TypedDict
class pc_state(TypedDict):
    input : str
    gen_outline : str
    gen_blog : str
     
```

```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
llm = ChatGroq(
    model="openai/gpt-oss-120b" ,
    temperature= 0.1
)
r = llm.invoke("when is pakistan independed")
print(r)
```

```python
def pc_processing(state : pc_state) -> pc_state :
    input = state['input']
    
    # now generate outline from the input
    prompt1 = f' generate the out line from the input {input}'
    
    # now we will give this  prompt1 to llm 
    gen_outline = state['gen_outline']
    gen_outline = llm.invoke(prompt1)
    
    #now generate a blog from the outline
    prompt2 = f'generate the blog from the foloowing outline \n {gen_outline}'
    
    #now give that prompt2 to llm
    gen_blog = state['gen_blog']
    gen_blog = llm.invoke(prompt2)
    
    # now update the state again
    state['gen_blog'] = gen_blog
    
    return state
```
